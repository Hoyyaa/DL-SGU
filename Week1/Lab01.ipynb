### Bài 6. Phép toán và nhân ma trận

```python
import torch
a = torch.randn(2, 3)
b = torch.randn(2, 3)
print("a:")
print(a)
print("b:")
print(b)
print("\na + b:")
print(a + b)
print("\na / b:")
print(a / b)
print("\na^2:")
print(a ** 2)
print("\nMatrix multiplication:")
print(a @ b.T)
```

- Phân biệt a * b và a @ b.T:

    - a @ b.T: Là phép nhân ma trận (matrix multiplication).  
- Shape của a @ b.T:
Vì a có shape $(2, 3)$ và b.T có shape $(3, 2)$, kết quả sẽ có shape là torch.Size([2, 2])

### Bài 8. Autograd cơ bản

```python
import torch
x = torch.tensor(
3.6,
requires_grad=True
)
y = x * x
y.backward()
print("x =", x)
print("y =", y)
print("x.grad =", x.grad)
```

- Đạo hàm lý thuyết $y' = 2x = 2 \times 3.6 = 7.2$, khớp hoàn toàn với kết quả Autograd. 
- Với hàm bậc 2 mới, $y' = 6x + 2 = 6(3.6) + 2 = 23.6$,Autograd tính chính xác giá trị đạo hàm. 

### Bài 9. Gradient Descent tìm hệ số đa thức

```python
import numpy as np
import torch

polynomial = np.poly1d([1, 2, 3])
N = 20

X = np.random.randn(N, 1) * 5
Y = polynomial(X)

XX = np.hstack([
    X * X,
    X,
    np.ones_like(X)
])

w = torch.randn(3, 1, requires_grad=True)
x = torch.tensor(XX, dtype=torch.float32)
y = torch.tensor(Y, dtype=torch.float32)

optimizer = torch.optim.NAdam([w], lr=0.01)

print("Initial coefficients:")
print(w)

for epoch in range(1000):
    optimizer.zero_grad()
    y_pred = x @ w
    mse = torch.mean(torch.square(y - y_pred))
    mse.backward()
    optimizer.step()

print("Learned coefficients:")
print(w)
```

Hệ số học được hội tụ chính xác về [1, 2, 3] của đa thức ban đầu.  
- optimizer.zero_grad() xóa bộ nhớ đệm đạo hàm cũ.  
- mse.backward() tính gradient thông qua lan truyền ngược.  
- optimizer.step() cập nhật trọng số w theo giải thuật NAdam. 

### Bài 10. Giải hệ phương trình bằng Autograd

```python
import random
import torch

A = torch.tensor(random.random(), requires_grad=True)
B = torch.tensor(random.random(), requires_grad=True)
C = torch.tensor(random.random(), requires_grad=True)
D = torch.tensor(random.random(), requires_grad=True)

EPOCHS = 2000
optimizer = torch.optim.NAdam([A, B, C, D], lr=0.01)

for _ in range(EPOCHS):
    y1 = A + B - 9
    y2 = C - D - 1
    y3 = A + C - 8
    y4 = B - D - 2

    sqerr = y1*y1 + y2*y2 + y3*y3 + y4*y4

    optimizer.zero_grad()
    sqerr.backward()
    optimizer.step()
    
print("A =", A)
print("B =", B)
print("C =", C)
print("D =", D)
```

- Hệ phương trình bậc nhất 4 ẩn này có phụ thuộc tuyến tính (thỏa mãn $y1 - y2 = y3 - y4$).
- Vì số phương trình độc lập ít hơn số ẩn nên hệ bài toán có vô số nghiệm. Tùy vào giá trị khởi tạo ban đầu mà mô hình sẽ tối ưu về các bộ nghiệm khác nhau.

### Bài 11. Viết chương trình PyTorch để thực hiện các yêu cầu sau:

- Tạo Tensor X có 24 phần tử và reshape lần lượt thành 4 × 6, 2 × 12 và 2 × 3 × 4.

- Tạo hai Tensor A, B kích thước 3 × 4; tính A+B, A*B, A/B, trung bình theo cột và độ lệch chuẩn theo cột.


- Cho y = 2x2 + 5x + 3. Dùng autograd tính dy/dx tại x = 2 và kiểm tra bằng phép tính thủ công.

- Thay đa thức trong Bài 9 bằng y = 2x2 - 3x + 5; huấn luyện và cho biết các hệ số tìm được.

```python
import numpy as np
import torch

print("=================== CÂU 1 ===================")
# 1. Tạo Tensor X có 24 phần tử (dùng arange hoặc randn)
X = torch.arange(24)
print("Tensor X ban đầu (24 phần tử):")
print(X)

# Reshape thành 4x6, 2x12 và 2x3x4
X_4x6 = X.reshape(4, 6)
X_2x12 = X.reshape(2, 12)
X_2x3x4 = X.reshape(2, 3, 4)

print("\nReshape 4x6:\n", X_4x6)
print("\nReshape 2x12:\n", X_2x12)
print("\nReshape 2x3x4:\n", X_2x3x4)


print("\n=================== CÂU 2 ===================")
# 2. Tạo hai Tensor A, B kích thước 3x4
A = torch.randn(3, 4)
B = torch.randn(3, 4)

print("Tensor A:\n", A)
print("Tensor B:\n", B)

# Tính các phép toán
print("\nA + B:\n", A + B)
print("\nA * B (element-wise):\n", A * B)
print("\nA / B (element-wise):\n", A / B)

# Trung bình và độ lệch chuẩn theo cột (dim=0) cho A và B
print("\n[A] Trung bình theo cột (dim=0):", torch.mean(A, dim=0))
print("[A] Độ lệch chuẩn theo cột (dim=0):", torch.std(A, dim=0))

print("\n[B] Trung bình theo cột (dim=0):", torch.mean(B, dim=0))
print("[B] Độ lệch chuẩn theo cột (dim=0):", torch.std(B, dim=0))


print("\n=================== CÂU 3 ===================")
# 3. Cho y = 2*x^2 + 5*x + 3. Dùng autograd tính dy/dx tại x = 2
x = torch.tensor(2.0, requires_grad=True)
y = 2 * x**2 + 5 * x + 3

# Tính gradient tự động
y.backward()

print("Giá trị x:", x.item())
print("Giá trị y:", y.item())
print("Gradient dy/dx từ PyTorch (x.grad):", x.grad.item())

# Kiểm tra thủ công:
# y = 2*x^2 + 5*x + 3  =>  dy/dx = 4*x + 5
# Tại x = 2:  dy/dx = 4*(2) + 5 = 13
manual_grad = 4 * 2 + 5
print(f"Kiểm tra thủ công: dy/dx = 4*(2) + 5 = {manual_grad}")
print("Kết quả trùng khớp!" if x.grad.item() == manual_grad else "Kết quả không trùng khớp!")


print("\n=================== CÂU 4 ===================")
# 4. Thay đa thức bài 9 bằng y = 2*x^2 - 3*x + 5
# Sinh dữ liệu từ đa thức mới
polynomial = np.poly1d([2, -3, 5])
N = 20
np.random.seed(42)  # Cố định seed cho kết quả ổn định
X_np = np.random.rand(N, 1) * 5
Y_np = polynomial(X_np)

# Chuẩn bị ma trận đặc trưng XX = [x^2, x, 1]
XX_np = np.hstack([X_np**2, X_np, np.ones_like(X_np)])

# Khởi tạo trọng số w (gồm 3 hệ số cần học)
w = torch.randn(3, 1, requires_grad=True, dtype=torch.float32)
x_tensor = torch.tensor(XX_np, dtype=torch.float32)
y_tensor = torch.tensor(Y_np, dtype=torch.float32)

# Sử dụng optimizer NAdam với learning rate thích hợp
optimizer = torch.optim.NAdam([w], lr=0.1)

print("Hệ số khởi tạo ban đầu:")
print(w.detach().numpy().flatten())

# Vòng lặp huấn luyện
epochs = 3000
for epoch in range(epochs):
    optimizer.zero_grad()
    y_pred = x_tensor @ w
    mse = torch.mean(torch.square(y_tensor - y_pred))
    mse.backward()
    optimizer.step()

learned_coeffs = w.detach().numpy().flatten()
print(f"\nHệ số học được sau {epochs} epoch:")
print(f"a = {learned_coeffs[0]:.4f}, b = {learned_coeffs[1]:.4f}, c = {learned_coeffs[2]:.4f}")
print("Hệ số thực tế ban đầu: [2, -3, 5]")
```

- Yêu cầu 3: Đạo hàm lý thuyết $\frac{dy}{dx} = 4x + 5$. Tại $x=2 \implies \frac{dy}{dx} = 4(2) + 5 = 13.0$, khớp tuyệt đối với Autograd.
- Yêu cầu 4: Thuật toán tối ưu NAdam hội tụ thành công về bộ hệ số gốc [2.0, -3.0, 5.0] sau 2500 epoch.
